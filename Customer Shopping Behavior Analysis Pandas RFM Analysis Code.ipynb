# For filling missing values using Median

```python
df['Review Rating'] = df.groupby('Category')["Review Rating"].transform(lambda x : x.fillna(x.median()))
```

# For verifying whether we can fill the missing values in Review Rating Colmun

```python
df.isnull().sum()
```

# For changing columns into snakecase(i.e all small case with underscore)

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns = {'purchase_amount_(usd)': 'purchase_amount'})
```

```python
df.columns
```

# Creating New Column

```python
labels = ['Young Adult', 'Adult', 'Middle-Aged','Senior']
df['age_group'] = pd.qcut(df['age'], q = 4, labels = labels) # here qcut = dividing a continuous variable into equal-sized bins based on sample quantiles or data rank
```

```python
df[['age', 'age_group']].head(10)
```

# create new column purchase_frequency_days

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```
